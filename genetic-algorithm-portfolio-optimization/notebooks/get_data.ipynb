```python
import yfinance as yf
import numpy as np

def fetch_real_data(tickers, start_date, end_date):
    try:
        # دانلود داده‌های تاریخی
        data = yf.download(tickers, start=start_date, end=end_date)['Adj Close']
        # محاسبه بازده روزانه
        returns_daily = data.pct_change().dropna()
        # میانگین بازده سالانه
        annual_returns = returns_daily.mean() * 252
        # ماتریس کوواریانس سالانه
        annual_cov_matrix = returns_daily.cov() * 252
        return annual_returns.values, annual_cov_matrix.values
    except Exception as e:
        print(f"خطا در دانلود داده‌ها: {e}")
        return None, None

def format_matrix_no_commas(matrix):
    return "\n".join(
        [" ".join(f"{value:.8f}" for value in row) for row in matrix]
    )

# تنظیمات
tickers = ['GC=F', 'SI=F', 'CL=F', 'PL=F', 'RUB=X', 'JPY=X']
start_date = '2020-01-01'
end_date = '2024-12-31'

# دریافت بازده و ماتریس کوواریانس
real_returns, real_risks = fetch_real_data(tickers, start_date, end_date)

if real_returns is not None and real_risks is not None:
    # فرمت‌دهی بازده و ماتریس
    returns_formatted = " ".join(f"{r:.8f}" for r in real_returns)
    risks_formatted = format_matrix_no_commas(real_risks)

    # نمایش خروجی
    print("Returns:")
    print(returns_formatted)
    print("\nRisk Covariance Matrix:")
    print(risks_formatted)
else:
    print("خطا در دریافت داده‌ها")
```

```python
import yfinance as yf
import numpy as np
from scipy.optimize import minimize

def fetch_stock_data(tickers, start_date, end_date):
    print("Fetching data from Yahoo Finance...")
    data = yf.download(tickers, start=start_date, end=end_date)['Adj Close']
    returns_daily = data.pct_change().dropna()  # بازده روزانه
    annual_returns = returns_daily.mean() * 252  # بازده سالانه
    annual_cov_matrix = returns_daily.cov() * 252  # ماتریس کوواریانس سالانه
    return annual_returns.values, annual_cov_matrix.values, tickers

def optimize_portfolio(returns, cov_matrix):
    num_assets = len(returns)
    initial_weights = np.ones(num_assets) / num_assets  # وزن اولیه مساوی
    bounds = [(0, 1) for _ in range(num_assets)]  # محدودیت وزن‌ها بین 0 و 1
    constraints = [{'type': 'eq', 'fun': lambda w: np.sum(w) - 1}]  # مجموع وزن‌ها = 1

    def neg_sharpe_ratio(weights):
        portfolio_return = np.dot(weights, returns)
        portfolio_risk = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))
        return -portfolio_return / portfolio_risk

    result = minimize(neg_sharpe_ratio, initial_weights, bounds=bounds, constraints=constraints)
    return result.x, -result.fun

if __name__ == "__main__":
    # تنظیمات
    tickers = ['AAPL', 'GOOGL', 'AMZN', 'MSFT', 'TSLA', 'META', 'NFLX', 'NVDA', 'JPM', 'V']  # لیست سهام
    start_date = "2020-01-01"
    end_date = "2023-12-31"

    try:
        real_returns, real_risks, tickers = fetch_stock_data(tickers, start_date, end_date)

        optimal_weights, optimal_sharpe_ratio = optimize_portfolio(real_returns, real_risks)
        
        print(f"\nOptimal Risk-to-Reward Ratio (Sharpe Ratio): {optimal_sharpe_ratio:.4f}")
    except Exception as e:
        print(f"Error fetching or processing data: {e}")
```
